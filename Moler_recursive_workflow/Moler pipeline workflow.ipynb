```python
from importlib import import_module, metadata
 
packages = [
    ("MoLeR", "molecule_generation", "molecule-generation"),
    ("RDKit", "rdkit", "rdkit"),
    ("TensorFlow", "tensorflow", "tensorflow"),
    ("NumPy", "numpy", "numpy"),
    ("Scikit-learn", "sklearn", "scikit-learn"),
    ("Protobuf", "google.protobuf", "protobuf"),
    ("DPU Utilities", "dpu_utils", "dpu-utils"),
    ("More Itertools", "more_itertools", "more-itertools"),
    ("TF2-GNN", "tf2_gnn", "tf2-gnn"),
]
 
print("MoLeR environment package check\n")
 
failed = []
 
for display_name, module_name, distribution_name in packages:
    try:
        module = import_module(module_name)
 
        try:
            version = metadata.version(distribution_name)
        except metadata.PackageNotFoundError:
            version = getattr(module, "__version__", "version unavailable")
 
        print(f"PASS  {display_name:<18} {version}")
 
    except Exception as error:
        print(f"FAIL  {display_name:<18} {type(error).__name__}: {error}")
        failed.append(display_name)
 
print("\nResult:")
 
if failed:
    print("Missing or broken packages:", failed)
else:
    print("All core MoLeR packages imported successfully.")
```

```python
import numpy
import networkx
import six
import openpyxl
 
from mordred import Calculator, descriptors
from rdkit import Chem
 
print("NumPy:", numpy.__version__)
print("NetworkX:", networkx.__version__)
print("Six:", six.__version__)
print("openpyxl:", openpyxl.__version__)
 
calculator = Calculator(descriptors, ignore_3D=True)
molecule = Chem.MolFromSmiles("CCO")
result = calculator(molecule)
 
print("Mordred works")
print("Number of descriptors:", len(calculator.descriptors))
print("Calculated values:", len(result))
```

```python
import shutil

import subprocess

import sys
 
xtb_executable = shutil.which("xtb")
 
print("Jupyter Python:", sys.executable)

print("xTB executable:", xtb_executable)
 
if xtb_executable is None:

    print("FAIL: Jupyter cannot currently find xTB.")

else:

    result = subprocess.run(

        [xtb_executable, "--version"],

        stdout=subprocess.PIPE,

        stderr=subprocess.PIPE,

        text=True,

        errors="replace"

    )
 
    print("Return code:", result.returncode)

    print("\nStandard output:")

    print(result.stdout or "[No standard output returned]")
 
    print("\nStandard error:")

    print(result.stderr or "[No standard error returned]")
 
```
